MapReduce 코드

```python
from multiprocessing import Pool
from collections import defaultdict

def mapper(line):
    fields = line.strip().split(",")
    if len(fields) > 3:
        publisher = fields[4]
        return (publisher, 1)

def reducer(mapped_data):
    result = defaultdict(int)
    for item in mapped_data:
        if item:
            key, value = item
            result[key] += value
    return result

if __name__ == "__main__":
    with open("books.csv", encoding="utf-8") as f:
        lines = f.readlines()[1:]  # header 제거

    with Pool(4) as p:
        mapped = p.map(mapper, lines)

    final_result = reducer(mapped)

    for k, v in list(final_result.items())[:10]:
        print(k, v)
```
